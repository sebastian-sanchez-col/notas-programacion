## Creación de listas

Las listas son conjuntos ordenados de elementos (números, cadenas, listas, etc). Las listas se delimitan por corchetes ([ ]) y los elementos se separan por comas.

Las listas pueden contener elementos del mismo tipo:

```python
dias_laborales: [str] = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"] # lista de cadenas: [str] 
numeros_irracionales: [float] = [3.1416, 2.7183] # lista de flotantes: [float]
# Recuerda con las listas tambien se puede asignar 
# cual es su tipo, en este caso [str] (lista de strings) o [float] (lista de flotantes) 
```

O pueden contener elementos de tipos distintos:

```python
fecha_de_encuentro = ["Viernes", 17, "diciembre", 2021]
```

## Operaciones con listas

¿Como obtenemos un elemento de una lista? Si conocemos el indice del elemento de la lista que queremos obtener lo podemos obtener de la siguiente forma:

```python
print(dias_laborales[3]) # Recuerda que los arreglos empiezan desde 0
```

En programación, las variables como las listas se pueden identificar como un arreglo los cuales representan un conjunto de elementos donde cada uno de ellos corresponde a una posición.

Por ejemplo un archivador o una carpeta para guardar documentos, cada uno de los espacios del mueble para guardar documentos o cada uno de los espacios de la carpeta tiene un espacio correspondiente al que lo podemos llamar, por ejemplo, espacio 1, espacio 2 y espacio 3. 

De esta forma funcionan las listas, cada uno de los espacios de las listas se representan por un indice empezando desde el número 0.

```python
print(dias_laborales[0]) # Primera posición o elemento de la lista
print(dias_laborales[1])
print(dias_laborales[2])
print(dias_laborales[3])
print(dias_laborales[4]) # Quinta posición o elemento de la lista
```

Al ejecutar el código anterior, podemos ver que se muestra por pantalla el contenido de cada una de las posiciones de la lista, por lo cual vemos como las listas o variables similares a los arreglos tienen su posición inicial en 0.

## Eliminar elementos de una lista

Para quitar un elemento de una lista podemos utilizar la función remove:

Nota: Utilizamos la función copy para no afectar la lista original.

```python
semana_laboral_cuatro_dias: [str] =  dias_laborales.copy()
semana_laboral_cuatro_dias.remove("Viernes")
print(semana_laboral_cuatro_dias)
```

También podemos utilizar una eliminación de elementos por indice, como lo hacemos a continuación:

```python
semana_laboral_cuatro_dias: [str] = dias_laborales.copy()
semana_laboral_cuatro_dias.pop(4)
print(semana_laboral_cuatro_dias)
```
