## Variables

### Cadenas de texto

En programación, una variable es un espacio de memoria donde se almacena un valor, los datos almacenados pueden tener distintos tipos de información y se recuperan cuando el desarrollador los necesite.

En el siguiente código almacenamos un valor en una variable:

```python
message = "Hola todos, soy una cadena de texto"
```

La variable del código anterior es de tipo String o cadena de texto, antes del simbolo igual "=" se asigna el nombre con el que se identifica la variable, luego de este simbolo se ingresa el valor a almacenar de la variable. 

En el siguiente recuadro ingresa otra variable con otro mensaje, recuerda la utilización de el simbolo igual "=" y la utilización de las comillas para la cadena de texto, ya que el lenguaje python interpreta que va a crear una variable con el valor y el nombre de la variable que asignes.

Nota: Recuerda ir compilando o ejecutando cada casilla de código y respetar la sintaxis (forma de escritura del código) del lenguaje de programación para que el programa funcione de forma adecuada.

Si queremos visualizar que valor se está almacenando dentro de la variable que creamos vamos a hacer uso de la función _print()_.

Una función contiene un fragmento de código que puede o no recibir valores o variables, es código contenido en una fracción de código que se ejecuta siempre que lo llamemos.

```python
print(message)
```
