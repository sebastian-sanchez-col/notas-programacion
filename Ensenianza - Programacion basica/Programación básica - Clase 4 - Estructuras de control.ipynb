## Estructuras de control - condicionales

Los condicionales utilizan los operadores (tema visto en la clase 2) para permitir ejecutar un fragmento de código o ejecutar otro fragmento de código si no se cumple dicha condición.

Por ejemplo:

```python
# Nota: Identación son los espacios que pueden existir entre el código y el inicio del renglón. Ejemplo:
#  #Comentario identado a un espacio
#    #Comentario identado a dos espacios

palabra = "cafe" # Puedes cambiar la palabra por otra
cantidad_caracteres = len(palabra) # La función len() permite encontrar la longitud

if cantidad_caracteres == 4: # Los condicionales se representan con la palabra if seguido de la condición a evaluar, en este caso que haya 4 caracteres en la variable cantidad_caracteres
  print("La palabra tiene 4 caracteres") # Si se cumple que tenga 4 caracteres después de los dos puntos del "if" se ejecuta el código antes del else o antes de la identación
else:
  print("La palabra tiene *más de* 4 caracteres") # Si *no* se cumple que tenga 4 caracteres después de los dos puntos del "else" se ejecuta el código antes de la identación
print("-- Fin del else --") # Siguiente linea, está linea inicia la identación desde el principio por lo cual no se incluye dentro del fragmento de codigo a ejecutar en el else
```

## Estructuras de control - ciclo for

El ciclo for permite hacer iteraciones o repeticiones. Estas repeticiones se realizan dado un número máximo de iteraciones.

```python
for indice in range(1,5): # Se va a repetir el código que esta dentro del ciclo for en el rango (1,5) (La función rango incluye el 1 pero no el 5, ejecuta el código para visualizar el comportamiento)
  print("Iteracion: ", indice)
```

De forma similar podemos utilizar el ciclo for en python para mostrar el contenido de una lista utilizada previamente:

```python
for dia in dias_laborales:
  print(dia)
```

Podemos mezclar a su vez el ciclo for con un condicional

```python
for dia in dias_laborales: # Recorremos cada elemento en la lista
  if dia == "Martes": # Usamos como condición que el elemento sea igual a "Martes"
    print("La lista contiene el día martes") # Si el elemento es "Martes" se imprime el mensaje
```

## Estructuras de control - ciclo while

El ciclo while permite repetir un bloque de código mientras se cumpla una determinada condición. En este caso, condicion es una expresión booleana que se evalúa en cada iteración del ciclo (iteración: cada vez que se repite el código dentro del ciclo).

```python
while condicion:
    # bloque de codigo a repetir
```

Es importante asegurarse de que la condición en el while eventualmente se vuelva falsa, de lo contrario el ciclo se ejecutará infinitamente.

A continuación veremos un ejemplo de ciclo while donde si el usuario no ingresa la palabra "done" no termina de guardarse los datos.

En el ultimo for imprimos el contenido de la lista

```python
ingredients = []
ingredient = ""

# Use a while loop to prompt the user to input ingredients
while ingredient != "done": # If the user enters "done", break out of the loop
    ingredient = input("Enter a food ingredient (or 'done' to finish): ") # Input function allow us to register a new word
        
    # Otherwise, append the ingredient to the list
    ingredients.append(ingredient) # we are using append function to insert this element at list

print("Here are the ingredients you entered:")
for ingredient in ingredients: 
    print(ingredient)
```

## Estructura de control - try except 

Se utiliza para manejar errores y excepciones en un programa. Si se produce un error en un bloque de código try, se ejecuta un bloque de código alternativo except.

En el siguiente ejemplo, si el usuario ingresa un valor que no se puede convertir a entero (por ejemplo, una cadena de texto), se produce un error de tipo ValueError. 

Para manejar este error, se utiliza un bloque try/except. Si se produce un error de tipo ValueError, se ejecuta el bloque de código except y se imprime un mensaje de error. Si no se produce ningún error, se imprime el número ingresado.

```python
# Pedir al usuario que ingrese un número
numero_str = input("Ingrese un número: ")

# Intentar convertir el número a un entero
try:
    numero_int = int(numero_str)
    print("El número ingresado es:", numero_int)
except ValueError:
    print("El valor ingresado no es un número entero.") 
```

Otro ejemplo que podemos ver es el de dividir un numero entre 0, si lo dividimos entre 0 se ejecuta la excepción:

```python
# Solicitar al usuario que ingrese un número
numero_str = input("Ingresa un número: ")

# Intentar convertir el número a entero y dividirlo por 2
try:
    numero = int(numero_str)
    resultado = numero / 2
    print("La mitad de", numero, "es", resultado)
    
# Manejar el error de división por cero
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero")
    
# Manejar otros errores
except:
    print("Error: Algo salió mal")
```
