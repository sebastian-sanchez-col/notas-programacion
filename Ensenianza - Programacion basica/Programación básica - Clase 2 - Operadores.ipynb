## Operadores de suma, resta, división y multiplicación

Vimos en la anterior clase el uso de la suma junto a la función (fragmento de código) _print()_. Vamos a continuar haciendo uso de variables y los operadores de resta, división y multiplicación. En el siguiente ejemplo vamos a restar dos numeros

```python
primer_numero: int = 3 
segundo_numero: float = 1.223 # Como vimos en la anterior lección es posible utilizar variables decimales llamadas flotantes

print(primer_numero - segundo_numero) # Para restar los dos numeros utilizamos el símbolo de guión "-"
# Nota: recuerda que puedes colocar comentarios a tu código utilizando el símbolo "#"
```

De forma similar a la suma o resta se puede utilizar los operadores matematicos para calcular la división o multiplicación:

```python
tercer_numero: float = 3.2323
cuarto_numero: int = 3

resultado: float = tercer_numero / cuarto_numero # Recuerda que las variables guardan valores, 
                                                 # en este caso creamos una variable para guardar el resultado de la división
print(resultado)
```

## Operadores para comparación

Necesitamos comparar dos valores para conocer si se cumple una condición, por ejemplo tengo más manzanas en una variable o en otra?:

```python
manzanas_verdes: int = 3
manzanas_rojas: int = 5

print(manzanas_rojas > manzanas_verdes)
```

Como vemos al comparar si la variable _manzanas_rojas_ es mayor a la variable _manzanas_verdes_ el resultado es True (verdadero) con lo que hicimos una comparación entre dos variables.

De igual forma podemos comparar si dos variables son iguales:

```python
cantidad_de_monedas: int = 4
cantidad_de_billetes: int = 4

print(cantidad_de_monedas == cantidad_de_billetes)
```

Como podemos observar el operador == permite visualizar si el contenido de dos variables tienen el mismo valor.

De forma similar aplica el operador diferente de "!=" en el siguiente espacio para identificar si las variables _cantidad_de_monedas_ y _cantidad_de_billetes_ son diferentes:

Es dos mayor a tres?:

```python
2 > 3 # Operador mayor que
```

O es dos menor que tres?:

```python
2 < 3 # Operador menor que
```

Es cuatro mayor igual que cuatro?

```python
4 >= 4 # Operador mayor igual que
```

El operador mayor igual, nos permite identificar si dos valores son **o** mayores **o** iguales.

¿Qué pasaría si necesitamos que se cumplan dos condiciones al tiempo?

```python
print (3 > 2 and 4 > 2)
```

Como vemos las dos variables booleanas son verdaderas y al utilizar el operador logico _and_ si las dos variables son verdaderas se cumple toda la condición, de lo contrario si una de ellas es falsa toda la condición es falsa: 

```python
1 > 2 and 4 > 2
```

De forma contraria el operador _or_ en caso de tener una condición verdadera pero una condición falsa, toda la condición es verdadera:

```python
print(3 > 3 or 4 > 2)
```
